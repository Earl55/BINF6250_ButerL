```python
def read_fasta(path):

    names = []
    seqs = []
    with open(path) as fh:
        cur_name = None
        cur_seq = []
        for line in fh:
            line=line.strip()
            if not line: continue
            if line.startswith('>'):
                if cur_name is not None:
                    seqs.append(''.join(cur_seq))
                    names.append(cur_name)
                cur_name = line[1:].strip()
                cur_seq = []
            else:
                cur_seq.append(line.upper())
        if cur_name is not None:
            seqs.append(''.join(cur_seq))
            names.append(cur_name)
    return seqs, names
```

```python
seqs, names = read_fasta("./phmm_train_motif1.fasta")
```

```python
profile_hmm = ProfileHMM(sequences=seqs)
```
